## Step 1 - Write Python Code

Imagine you're writing instructions for a robot.

```python
name = "Rohan"
age = 5
print(name)
print(age)
```

This is stored as a `.py` file.

````python
name="Rohan"
age=5
print(name)
print(age)
````

## Step 2 - Python Interpreter

The Python Interpreter is a software program written mostly in C

When you run:

```bash
python example.py
```

the Python Interpreter starts.

Its job is to read, understand and execute your code.

example.py

↓

Reads Line 1

↓

Reads Line 2

↓

Reads Line 3

↓

Reads Line 4

## Step 3 - Lexer

The lexer breaks your code into small pieces called **tokens**.

Example:

```python
print("Hello")
```

Tokens:

- print
- (
- "Hello"
- )


## Step 4 - Parser

The parser checks whether your code follows Python grammar.

Example of a syntax error:

```python
print("Hello"
```

Missing `)` causes a `SyntaxError`.


## Step 5 - Abstract Syntax Tree (AST)

Python converts valid code into a tree structure called the **AST**.

Example:

```
    =
   / \
  x   +
     / \
   10 20
```


````python
import ast

tree = ast.parse("x = 10 + 20")
print(ast.dump(tree, indent=4))
````

## Step 6 - Compilation

Python compiles the AST into **Bytecode**.

Bytecode is not machine code.

It is a lower-level instruction set understood by the Python Virtual Machine.


## Step 7 - View Bytecode

Python lets us inspect bytecode using the `dis` module.


````python
import dis

def add(a,b):
    return a+b

dis.dis(add)
````

## Step 8 - Python Virtual Machine

The PVM reads bytecode and executes instructions one by one.


## Step 9 - Memory

Variables are stored in RAM while the program runs.


````python
name="Rohan"
age=5

print(id(name))
print(id(age))
````

## Step 10 - Output

After execution, Python displays the output on the console.


````python
print("Hello Python!")
````

## Step 11 - Garbage Collection

When objects are no longer needed, Python automatically frees memory.


````python
import gc

print(gc.isenabled())
````

### What happens internally when Python executes a program


"Python reads the source code, tokenizes it with the lexer, validates syntax using the parser, builds an Abstract Syntax Tree (AST), compiles it into bytecode, executes the bytecode with the Python Virtual Machine (PVM), uses the CPU to perform operations, stores objects in memory during execution, displays the output, and finally frees unused memory through garbage collection."
