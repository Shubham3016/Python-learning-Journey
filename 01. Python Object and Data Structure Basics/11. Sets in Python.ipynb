### Remove item

To remove an item in a set, use the **remove()**, or the **discard()** method.

```python
# Remove value from set

ac={10,20,203,30,430}
print(ac)

ac.remove(10)
print(ac)
```

```python
# remove throws an error if th evalue is missing

ac.remove(100)
print(ac)
```

```python
# create a set

thisset = {"apple", "banana", "cherry"}

# remove the element banana

thisset.remove("banana")
```

```python
print(thisset)
```

```python
# create a set

thisset = {"1", "2", "3"}

# discard the element 3

thisset.discard("3")

#show

print(thisset)

thisset.discard("10")

print(thisset)
```

## Join Two Sets
There are several ways to join two or more sets in Python.

You can use the `union()` method that returns a new set containing all items from both sets, or the `update()` method that inserts all the items from one set into another:

```python
set1 = {"a", "b" , "c"}
```

```python
set2 = {1, 2, 3}
```

```python
set3 = set1.union(set2)
```

```python
print(set3)
```

```python
a={10,20,30,40}
b={30,40,50,60}

print(a.union(b))
```

```python
print(a|b)
```

```python
print(a.intersection (b))

print( a & b)
# it ewill give value that appers in both list
```

```python
print(a.difference(b))

print(a-b)
print(b-a)
```

```python
print(a.symmetric_difference(b)) 
```

## The set() Constructor

It is also possible to use the set() constructor to make a set.

```python
set1 = set(("apple", "banana", "cherry")) # note the double round-brackets
```

```python
print(set1)
```

```python
a={10,20,30,40}
b={30,40,50,60}

print(a.issubset(b))
```

```python
a={30,40}
b={30,40,50,60}


print(a.issubset(b))
print(b.issuperset(a))

print(a.isdisjoint(b))
print(b.issubset(a))
# as 30 and 40 are in set a
```
