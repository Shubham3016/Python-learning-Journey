## Identity Operators

**is** Returns True if two variables refer to the same object in memory

What is Identity?

Imagine you and your friend each have a notebook.

Case 1: Same Notebook

Both of you are holding the exact same notebook

  

```python
x=1
y=2
print(x is y)
```

```python
x=3
y=3

print(x is y)
print(id (x))
print(id (y))

print(x==y)
```

```python
a="abc"
b="abc"
print(a is b)
print(id(a),id(b))
```

```python

x=["a","b","c"]
y=["a","b","c"]

print(x==y)
```

```python
x=["a","b","c"]
y=["a","b","c"]

print(x is y)

# is checks OBJECT IDENTITY
```

#### Because Python created two different list objects.



Different memory locations.

```python
x = [10,20]

y = x

print(x is y)
```

#### Both variables point to the same list.



```python
x = [10,20]

y = [10,20]

print(x is y)

print(id(x))
print(id(y))

# Same values Different objects
```

```python
customer = {
    "name":"John"
}

customer_backup = customer

print(customer_backup is customer)

print(id(customer_backup))
print(id(customer))
```

```python
customer_copy = {
    "name":"John"
}

customer==customer_copy

print(customer is customer_copy)

print(id(customer_copy))
print(id(customer))
```

```python
value = None

if value is None:
    print("No value")

    # Using is None is the Python style recommended by PEP 8 because None is a singleton (only one None object exists).
```

```python
a = [1, 2]
b = a
c=[1,2]
print(a)
print(b)

print(a==b) #a == b → Same values (True)
print(a is b) #a is b → Same object (True)
print(a==c) #a == c → Same values (True)
print(a is c) #a is c → Different list objects (False)
print(a is not c) #a is not c → Different objects (True)
```

#### TIPS
Use == when you want to compare values.

Use is when you want to check object identity, especially with None.

Avoid using is to compare numbers, strings, or other immutable values, because Python may reuse some objects internally (called interning), which can make results appear inconsistent.

Membership Operators are used to check whether a value exists inside a sequence or collection.

```python

text="Hello World"

print("H" in text)
print("hello" in text)
print("z" not in text)
```

```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
```
