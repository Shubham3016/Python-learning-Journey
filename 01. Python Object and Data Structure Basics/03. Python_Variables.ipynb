## 1️⃣ What is a Variable?

A **variable** is a **name that refers to a value**.

Think of it like a **label on a toy box**.

````python
name = "Bob"
age = 27

print(name)
print(age)
````


### Explain Like Beginner 👶

Imagine three toy boxes.

```
+-------------+
| Cars        |
+-------------+

+-------------+
| Chocolates  |
+-------------+

+-------------+
| Balls       |
+-------------+
```

Each box has a **name**.

Python variables work exactly the same way.

## 4️⃣ Common Data Types

````python
name = "Alice"      # str
age = 25            # int
height = 5.8        # float
is_student = True   # bool
marks = [90, 95]    # list

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
print(type(marks))
````

## 5️⃣ Reassigning Variables

````python
score = 10
print(score)

score = 20
print(score)
````

````python
# Execrice

print("info@datawithbaraa.com")
print("support@datawithbaraa.com")
print("WWW.datawithbaraa.com")

##need to print it using dynamic approach

print("\n")
a='datawithbaraa.com'
print("info@"+a)
print("support@"+a)
print("WWW."+a)

print("\n")
a1="info"
a2="support"
a3="WWW"
domain='datawithbaraa.com'
print(a1+"@"+domain)
print(a2+"@"+domain)
print(a3+domain)
````
