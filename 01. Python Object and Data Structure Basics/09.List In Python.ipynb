### Rest Collector *

```python
person=["MAria",29,"DE","India",'Pune']
print(person)

name,age,role,country,city=person


print(name);print(age);print(role);print(country)

# instead of calling all variables we can have Rest(*) collector

name,*details,country=person

print(name)
print(details)
print(city)

print(name,details)


```
