## What is a Data Type?

Imagine you have different boxes.

🍎 Apple Box → Fruits
🚗 Car Box → Cars
📚 Book Box → Books

Python also needs different boxes for different kinds of information.

These boxes are called **Data Types**.


````python
name="Alice"
age=28
salary=75000.50
is_employee=True

print(name,type(name))
print(age,type(age))
print(salary,type(salary))
print(is_employee,type(is_employee))
````

## Numeric Types

Python stores numbers as int, float and complex.

````python
a=10
b=10.5
c=2+3j
print(type(a),type(b),type(c))
````

## Strings

Strings store text.

````python
company='OpenAI'
print(company)
print(company.upper())
````

## Boolean

Only True or False.

````python
is_logged_in=True
print(is_logged_in)
````

## Lists

Lists are ordered and mutable.

````python
fruits=['Apple','Banana','Mango']
fruits.append('Orange')
print(fruits)
````

## Tuple

Tuples are ordered and immutable.

````python
days=('Mon','Tue','Wed')
print(days)
````

## Set

Sets keep unique values.

````python
nums={1,2,2,3}
print(nums)
````

## Dictionary

Stores key-value pairs.

````python
employee={'name':'Rahul','age':30}
print(employee['name'])
````

## None

Represents no value.

````python
data=None
print(data)
print(type(data))
````

## Type Conversion

Convert one type into another.

````python
age='28'
print(int(age)+2)
````

# Real Company Example

An e-commerce website may store:

- Customer Name → string
- Customer Age → int
- Product Price → float
- Premium Member → bool
- Cart Items → list
- Product Details → dictionary


````python

customer={
    "name":"Amit",
    "age":29,
    "premium":True,
    "cart":["Laptop","Mouse"],
    "wallet":1200.75
}

for k,v in customer.items():
    print(k,"=>",v,type(v))
````

## Module

A module is a Python file (.py) that contains reusable code such as functions, variables, classes, and constants. Modules help you organize code and avoid rewriting the same logic.

Imagine you have a toolbox.

Instead of carrying every tool in your pocket, you keep them in separate boxes.

🧰 Toolbox

-  🪚 Saw Box
-  🔩 Wrench Box

Whenever you need a tool, you open the correct box.

Python works the same way.

Instead of writing everything in one file, Python keeps related code in separate modules.
