 ## 3. Indexing SLicing 


Each character has an index.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

### In python we use the concept of indexing and we use [] square brackets.


````python
text="Python"
print(text[0])
print(text[-1])
````



Syntax: `string[start:end:step]`

Example:
```
Python
012345
```
        Positive Index

        0   1   2   3   4   5

        P   y   t   h   o   n

       -6  -5  -4  -3  -2  -1

        Negative Index


#### We can use a `:` to perform slicing which grabs everything up to a designated point. For example:

````python
text="Python"

print(text[0:3]) 

print(text[2:])

print(text[::-1])

print(text[::])
````

````python
mystring="Believe yourself"
print(mystring)

# using indexing print first charcter B
 
print(mystring[0])
print(mystring[-0])
print(mystring[-16])

# using indexing print s

print(mystring[12])
print(mystring[-4])
````

````python
alpha="abcdefgh"
print(alpha)

print(alpha[:3])
print(alpha[0:3])
print(alpha[-8:-5])
print(alpha[:-5])

````

````python

print(alpha[3:])
print(alpha[-5:])

# print the string upto
#print the sting b to f

print(alpha[1:6])
print(alpha[-7:-2])

# string indexing using steps

print(alpha[:])

print(alpha[::])

print(alpha)
````

````python
### Indexing using steps
alpha="abcdefgh"

print(alpha[3:6])
print(alpha[0::1])
print(alpha[::1])

# step 2 skip 

#start:stop:step

print(alpha[0:7:2])
print(alpha[::2])

print(alpha[0::3])

print(alpha[::-1])
````

````python
date="2025-04-09"
print(date)

# Extract year
print(date[:4])
print(date[0:4])

# MOnth
print(date[5:7])

# Day
print(date[8:])
````

````python

# Returns True if all characters in the string are alphanumeric
# The isalnum() method returns True if all the characters are alphanumeric, meaning alphabet letter (a-z) and numbers (0-9).

a="ddsd22w1"
print(a.isalnum())


# Returns True if all characters in the string are in the alphabet
#The isalpha() method returns True if all the characters are alphabet letters (a-z).

print(a.isalpha())

````

````python

# The isidentifier() method returns True if the string is a valid identifier, otherwise False.

# A string is considered a valid identifier if it only contains alphanumeric letters (a-z) and (0-9), or underscores (_). 
#A valid identifier cannot start with a number, or contain any spaces.
    
b="1221Axsds"
print(b.isidentifier()) 
````
